```python
from sklearn.metrics import silhouette_score
```

```python
silhouette_values = []
```

```python
for k in range(2, 9):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    cluster_labels = kmeans.fit_predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        cluster_labels
    )

    silhouette_values.append(score)
```

```python
k_values = list(range(2, 9))

for k, score in zip(k_values, silhouette_values):
    print("K =", k, "Silhouette Score =", score)
```

```python
plt.plot(k_values, silhouette_values, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score")
plt.show()
```

```python
def segment_recommendation(row):
    
    recommendations = []
    
    if row["Investment Buyers (%)"] >= 60:
        recommendations.append(
            "Investment-focused property marketing"
        )
    
    if row["Loan Application Rate (%)"] >= 50:
        recommendations.append(
            "Financing and loan support"
        )
    
    if row["Average_Satisfaction"] < df["satisfaction_score"].mean():
        recommendations.append(
            "Customer experience improvement"
        )
    
    if not recommendations:
        recommendations.append(
            "General personalized marketing"
        )
    
    return "; ".join(recommendations)
```

```python
master_segment_summary["Recommended Strategy"] = (
    master_segment_summary.apply(
        segment_recommendation,
        axis=1
    )
)
```
